```python
us_df["country"] = "USA"
gb_df["country"] = "Great Britain"
in_df["country"] = "India"
```

Combined DataFrame

```python
df = pd.concat(
    [us_df, gb_df, in_df],
    ignore_index=True
)
```

Create Array

```python
views = df["views"].to_numpy()
likes = df["likes"].to_numpy()
dislikes = df["dislikes"].to_numpy()
comments = df["comment_count"].to_numpy()
```

```python
arrays = {
    "views": views,
    "likes": likes,
    "dislikes": dislikes,
    "comments": comments
}

for name, arr in arrays.items():
    print(f"{name}:")
    print("Shape:", arr.shape)
    print("Data type:", arr.dtype)
    print("-------------------")
```

Top 10 views of combined Data

```python
top_10_views = np.sort(views)[-10:][::-1]

print("10 Largest View Counts:")
print(top_10_views)
```

```python

print("Likes:", likes[10:20])
print("Dislikes:", dislikes[10:20])
print("Comments:", comments[10:20])
```

Calculate Like Ratio

```python
like_ratio = np.divide(
    likes,
    likes + dislikes,
    out=np.zeros(len(likes), dtype=float),
    where=(likes + dislikes) != 0
)
like_ratio
```

Calculate Comment_rate

```python
comment_rate = np.divide(
    comments,
    views,
    out=np.zeros(len(comments), dtype=float),
    where=views != 0
)
comment_rate
```

```python
print(views.min(), views.max())
```

Calculate normalized_views

```python
normalized_views = (
    (views - views.min()) /
    (views.max() - views.min())
)
print(normalized_views)
```

Calculate normalized_comments

```python
normalized_comments = (
    (comments - comments.min()) /
    (comments.max() - comments.min())
)
print(normalized_comments)
```

Estimagte the Engagement score

```python
engagement_score = (
    0.5 * like_ratio +
    0.3 * normalized_views +
    0.2 * normalized_comments
)
print(engagement_score.min(), engagement_score.max())
```

Find average views by category

```python
category_analysis = df.groupby(
    by="category_name",
    level=0,
).agg(
    total_views=("views", "sum"),
    average_views=("views", "mean"),
    average_likes=("likes", "mean")
)

print(category_analysis)
```

Trending lifespan
Suppose one video appeared on the trending list on five different days.

Its trending lifespan is five days

This is an important business insight because it helps identify videos that remain popular for lon

```python
lifespan = df.groupby(
    ["country", "video_id"]
)["trending_date"].nunique()

print(lifespan.sort_values(ascending=False).head(10))
```
