```python


import pandas as pd
import os

country_files = {
    "USvideos": "USA",
    "GBvideos": "Great Britain",
    "INvideos": "India",
    "FRvideos": "France"
}

folder_path = r"C:\Users\Sasirekha\OneDrive\Desktop\TrendPulse\Masai_Mini_Project\dataset"

dataframes = []

for file, country in country_files.items():

    file_path = os.path.join(folder_path, file + ".csv")
    if not os.path.isfile(file_path):
        print(f"Skipping {country}: file not found at {file_path}")
        continue

    df = pd.read_csv(file_path)

    df["country"] = country

    dataframes.append(df)

    print(f"{country}: {len(df)} records loaded")

combined_df = pd.concat(dataframes, ignore_index=True)

print("Total records:", len(combined_df))
```
