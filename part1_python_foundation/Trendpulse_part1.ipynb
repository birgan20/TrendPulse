Understand the concepts:

def – Creates a function.
path – File location.
country – Country name.
max_rows=None – Default argument.
return – Sends the result back.
to_dict("records") – Converts DataFrame rows into dictionaries.

```python
def load_country(path,country, max_rows = None):

  df = pd.read_csv(path, nrows = max_rows)

  df["country"] = country

  required_columns = [
      'video_id',
      'title',
      'views',
      'likes',
      'dislikes',
      'comment_count'
  ]
  missing = set(required_columns) - set(df.columns)

  if missing:
      raise ValueError(f"Missing columns: {missing}")

  records = df.to_dict("records")

  print(f"{country}: {len(records)} records loaded")

  return records
```

### 2. Title Features Using Python Loops

The `title_features()` function analyses YouTube video titles.

We use:
- `for` to iterate through title characters and words.
- `range()` to access characters using their index positions.
- `while` to process words until the end of the list.
- `continue` to skip empty words.
- `break` to stop processing when a maximum word limit is reached.

```python
def title_features(title):

    words = title.split()

    uppercase_words = 0
    
    word_count = len(words)
     # 3. WHILE LOOP: Process words
    i = 0

    while i < len(words):

        word = words[i]

        # 4. CONTINUE: Skip empty words
        if word == "":
            i += 1
            continue

        # 5. BREAK: Stop after 20 words
        if word_count == 20:
            break

    for word in words:

        if word.isupper():
            uppercase_words += 1

    special_marks = 0

    for character in title:

        if character in "!?":
            special_marks += 1

    return {
        "uppercase_words": uppercase_words,
        "special_marks": special_marks,
        "word_count": len(words)
    }
```

```python
print(
    title_features("Final Robots Review?!")
)
```

classify videos into five categories:
- Viral Hit
- Steady Performer
- Divisive
- Low Engagement
- Unrated

```python
def classify_video(record):

    views = record["views"]
    likes = record["likes"]
    dislikes = record["dislikes"]

    if record.get("ratings_disabled", False):
        return "Unrated"

    if likes + dislikes == 0:
        return "Unrated"

    like_ratio = likes / (likes + dislikes)

    if views >= 1000000 and like_ratio >= 0.90:
        return "Viral Hit"

    elif views >= 100000 and like_ratio >= 0.75:
        return "Steady Performer"

    elif like_ratio < 0.50:
        return "Divisive"

    else:
        return "Low Engagement"
```
